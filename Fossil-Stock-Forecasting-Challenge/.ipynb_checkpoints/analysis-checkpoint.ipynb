```python
values = pd.DataFrame(train_df[train_df.sku_name=="ABEANHARLE"].sellin.values)
dataframe = pd.concat([values.shift(2), values.shift(1), values], axis=1)
dataframe.columns = ['t-2', 't-1', 't']
result = dataframe.corr()
print(result)
```
