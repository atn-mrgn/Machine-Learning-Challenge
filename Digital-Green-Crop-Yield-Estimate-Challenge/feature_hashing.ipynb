```python
from sklearn.feature_extraction import FeatureHasher
from sklearn.pipeline import FeatureUnion
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import FunctionTransformer
import pandas as pd
import numpy as np

# Supposons que tu as un DataFrame df avec des variables catégorielles et une variable numérique
# Exemple de données factices
data = {
    'num_var': [10, 20, 15, 25, 30, 18],
    'cat_var1': ['A', 'B', 'A', 'C', 'B', 'C'],
    'cat_var2': ['X', 'Y', 'Y', 'X', 'X', 'Y'],
    'cat_var3': ['M', 'N', 'M', 'N', 'M', 'N']
}
df = pd.DataFrame(data)

# Liste des variables catégorielles à hasher et des colonnes numériques à conserver
cat_cols_to_hash = ['cat_var1', 'cat_var2', 'cat_var3']
num_cols_to_keep = ['num_var']

# Création d'un transformateur pour les variables catégorielles
cat_transformers = [('hasher_' + col, FeatureHasher(n_features=5, input_type='string'), col) for col in cat_cols_to_hash]

# Création d'un ColumnTransformer pour traiter sélectivement les variables catégorielles et numériques
ct = ColumnTransformer([
    ('categorical', FeatureUnion(cat_transformers), cat_cols_to_hash),
    ('numeric', 'passthrough', num_cols_to_keep)
])

# Transformation des variables catégorielles et numériques avec le ColumnTransformer
transformed_features = ct.fit_transform(df)

# Création d'un DataFrame avec les résultats transformés
columns = [f'hashed_{col}' for col in cat_cols_to_hash] + num_cols_to_keep
hashed_df = pd.DataFrame(transformed_features, columns=columns)
print(hashed_df)
```
