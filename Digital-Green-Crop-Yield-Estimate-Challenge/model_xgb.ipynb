```python
import pandas as pd
import xgboost as xgb
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split

# Supposons que tu as un DataFrame df avec des variables catégorielles et une variable cible 'target'
# Exemple de données factices
data = {
    'cat_var1': ['A', 'B', 'A', 'C', 'B', 'C'],
    'cat_var2': ['X', 'Y', 'Y', 'X', 'X', 'Y'],
    'target': [1, 0, 1, 1, 0, 0]
}
df = pd.DataFrame(data)

# Encodage One-Hot pour les variables catégorielles
encoder = OneHotEncoder(sparse=False, handle_unknown='ignore')
encoded_cat_vars = encoder.fit_transform(df[['cat_var1', 'cat_var2']])

# Création d'un nouveau DataFrame avec les variables encodées
df_encoded = pd.concat([df.drop(columns=['cat_var1', 'cat_var2']), pd.DataFrame(encoded_cat_vars)], axis=1)

# Séparation en ensembles d'entraînement et de test
train, test = train_test_split(df_encoded, test_size=0.2, random_state=42)

# Gestion des modalités absentes dans la base de test
missing_cols = set(train.columns) - set(test.columns)
for col in missing_cols:
    test[col] = 0  # Remplacer les modalités manquantes par 0 dans la base de test

# Préparation des données pour XGBoost
X_train = train.drop(columns=['target'])
y_train = train['target']
X_test = test.drop(columns=['target'])
y_test = test['target']

# Entraînement du modèle XGBoost
model = xgb.XGBClassifier()
model.fit(X_train, y_train)

# Évaluation du modèle
accuracy = model.score(X_test, y_test)
print(f"Accuracy: {accuracy}")
```
