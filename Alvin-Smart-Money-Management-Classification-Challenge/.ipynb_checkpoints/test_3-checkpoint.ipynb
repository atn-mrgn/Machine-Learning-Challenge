```python
import pandas as pd
import numpy as np
import seaborn as sns
from catboost import Pool, CatBoostClassifier
import matplotlib.pyplot as plt
from lightgbm import LGBMClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, log_loss
from sklearn.impute import SimpleImputer
```

```python
train_df=pd.read_csv("Train.csv")
extra=pd.read_csv("extra_data.csv")

log_PURCHASE_VALUE=np.log(train_df.PURCHASE_VALUE)
bins=[np.quantile(log_PURCHASE_VALUE, i) for i in np.linspace(0,1,12)]
train_df["Var_1"]=pd.cut(log_PURCHASE_VALUE, bins).astype("object")

freq_dict = pd.concat([extra.MERCHANT_NAME, train_df.MERCHANT_NAME], axis=0).reset_index(drop=True).value_counts().to_dict()
category={j:f"cat_{i}" for i,j in enumerate(np.unique(list(freq_dict.values())))}
MERCHANT_NAME_code={key:category[val] for key, val in freq_dict.items()}
#train_df["Var_2"]=train_df.MERCHANT_NAME.map(MERCHANT_NAME_code)

per_freq_dict = pd.concat([extra.MERCHANT_NAME, train_df.MERCHANT_NAME], axis=0).reset_index(drop=True).value_counts(normalize=True).to_dict()
#train_df["Var_3"]=train_df.MERCHANT_NAME.map(per_freq_dict)

true_columns=["MERCHANT_CATEGORIZED_AT","PURCHASED_AT","Transaction_ID","USER_AGE","MERCHANT_NAME","PURCHASE_VALUE",'USER_GENDER']
df=train_df.drop(true_columns, axis=1)
val_columns=[i for i in train_df.MERCHANT_CATEGORIZED_AS.unique() if (train_df.MERCHANT_CATEGORIZED_AS[train_df.MERCHANT_CATEGORIZED_AS==i].shape[0]/train_df.shape[0])>0.01]
df=df[df.MERCHANT_CATEGORIZED_AS.isin(val_columns)]

target=sorted(df.MERCHANT_CATEGORIZED_AS.unique())

cat_codes={
    False:1,
    True:0,
    "Male":1,
    "Female":0,
}
#df.USER_GENDER=df.USER_GENDER.map(cat_codes)
df.IS_PURCHASE_PAID_VIA_MPESA_SEND_MONEY=df.IS_PURCHASE_PAID_VIA_MPESA_SEND_MONEY.map(cat_codes)
encoder=LabelEncoder()
df.MERCHANT_CATEGORIZED_AS=encoder.fit_transform(df.MERCHANT_CATEGORIZED_AS)
var_1_enc=LabelEncoder()
df.Var_1=var_1_enc.fit_transform(df.Var_1)
```

```python
X_train, X_test, y_train, y_test=train_test_split(df.drop(["MERCHANT_CATEGORIZED_AS"], axis=1), df.MERCHANT_CATEGORIZED_AS, test_size=.2 ,random_state=25)
```

```python
X_train.head(2)
```

```python
train_data=Pool(
    data=X_train,
    label=y_train,
    cat_features=["USER_ID"]
)
eval_data=Pool(
    data=X_test,
    label=y_test,
    cat_features=["USER_ID"]
)
```

```python
model = CatBoostClassifier(iterations=100,
                           learning_rate=0.14,
                           depth=8,
                           custom_metric=['MultiClass']
)
# Fit model
model.fit(
    train_data,
    eval_set=eval_data,
    verbose=True
)
# Get predicted classes
preds_class = model.predict(eval_data)
```

```python
print(model.get_best_score())
```

```python
preds_proba = model.predict_proba(eval_data)
# Get predicted RawFormulaVal
preds_raw = model.predict(eval_data,
                          prediction_type='RawFormulaVal')
```

```python
print(classification_report(y_test, preds_class))
print(f"Log_loss:{log_loss(y_test, preds_proba)}")
```

```python
test_df=pd.read_csv("Test.csv")
test_df["Var_1"]=pd.cut(np.log(test_df.PURCHASE_VALUE), bins)
#test_df["Var_2"]=test_df.MERCHANT_NAME.map(MERCHANT_NAME_code)
#test_df["Var_3"]=test_df.MERCHANT_NAME.map(per_freq_dict)
test_df.Var_1=var_1_enc.fit_transform(test_df.Var_1)
test_df.IS_PURCHASE_PAID_VIA_MPESA_SEND_MONEY=test_df.IS_PURCHASE_PAID_VIA_MPESA_SEND_MONEY.map(cat_codes)
test_df.drop(true_columns, axis=1, inplace=True)
predictions=pd.DataFrame(model.predict_proba(test_df), columns=target)
ss=pd.read_csv("SampleSubmission.csv")
for column in predictions.columns:
    ss[column]=predictions[column]

ss.head()
```

```python
ss.to_csv("submission_20.csv", index=False)
```

```python
extra_df=pd.read_csv("extra_data.csv")
extra=extra_df.copy()
extra_df.drop("MERCHANT_CATEGORIZED_AS", axis=1, inplace=True)
extra_df["Var_1"]=pd.cut(np.log(extra_df.PURCHASE_VALUE), bins)
#test_df["Var_2"]=test_df.MERCHANT_NAME.map(MERCHANT_NAME_code)
#test_df["Var_3"]=test_df.MERCHANT_NAME.map(per_freq_dict)
extra_df.Var_1=var_1_enc.fit_transform(extra_df.Var_1)
extra_df.IS_PURCHASE_PAID_VIA_MPESA_SEND_MONEY=extra_df.IS_PURCHASE_PAID_VIA_MPESA_SEND_MONEY.map(cat_codes)
extra_df.drop(true_columns, axis=1, inplace=True)
predictions=pd.DataFrame(model.predict_proba(extra_df), columns=target, index=extra.Transaction_ID)
```

```python
category=encoder.inverse_transform(model.predict(extra_df))
table=pd.DataFrame(predictions[predictions>.50].sum(axis=1), columns=["Score"])
table["Cat"]=category

sup_data=pd.concat([extra, table.reset_index(drop=True)], axis=1)
sup_data.MERCHANT_CATEGORIZED_AS=sup_data.Cat
sup_data=sup_data[sup_data.Score!=0]
sup_data=sup_data.drop(['Score',"Cat"], axis=1).reset_index(drop=True)
```
