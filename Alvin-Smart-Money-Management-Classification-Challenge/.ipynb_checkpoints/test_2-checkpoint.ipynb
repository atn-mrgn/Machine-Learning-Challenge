```python
import pandas as pd
import numpy as np
import seaborn as sns
import lightgbm as lgb
import matplotlib.pyplot as plt
from lightgbm import LGBMClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.metrics import classification_report, log_loss
```

```python
train_df=pd.read_csv("Train.csv")
extra=pd.read_csv("extra_data.csv")

log_PURCHASE_VALUE=np.log(train_df.PURCHASE_VALUE)
bins=[np.quantile(log_PURCHASE_VALUE, i) for i in np.linspace(0,1,12)]
train_df["Var_1"]=pd.cut(log_PURCHASE_VALUE, bins).astype("object")

freq_dict = pd.concat([extra.MERCHANT_NAME, train_df.MERCHANT_NAME], axis=0).reset_index(drop=True).value_counts().to_dict()
category={j:f"cat_{i}" for i,j in enumerate(np.unique(list(freq_dict.values())))}
MERCHANT_NAME_code={key:category[val] for key, val in freq_dict.items()}
train_df["Var_2"]=train_df.MERCHANT_NAME.map(MERCHANT_NAME_code)

per_freq_dict = pd.concat([extra.MERCHANT_NAME, train_df.MERCHANT_NAME], axis=0).reset_index(drop=True).value_counts(normalize=True).to_dict()
train_df["Var_3"]=train_df.MERCHANT_NAME.map(per_freq_dict)

true_columns=["MERCHANT_CATEGORIZED_AT","PURCHASED_AT","Transaction_ID","USER_AGE","USER_GENDER","PURCHASE_VALUE","MERCHANT_NAME"]
df=train_df.drop(true_columns, axis=1)
val_columns=[i for i in train_df.MERCHANT_CATEGORIZED_AS.unique() if (train_df.MERCHANT_CATEGORIZED_AS[train_df.MERCHANT_CATEGORIZED_AS==i].shape[0]/train_df.shape[0])>0.01]
df=df[df.MERCHANT_CATEGORIZED_AS.isin(val_columns)]

target=sorted(df.MERCHANT_CATEGORIZED_AS.unique())

cat_codes={
    False:1,
    True:0,
    "Male":1,
    "Female":0,
}
#df.USER_GENDER=df.USER_GENDER.map(cat_codes)
df.IS_PURCHASE_PAID_VIA_MPESA_SEND_MONEY=df.IS_PURCHASE_PAID_VIA_MPESA_SEND_MONEY.map(cat_codes)
df.MERCHANT_CATEGORIZED_AS=LabelEncoder().fit_transform(df.MERCHANT_CATEGORIZED_AS)
var_1_enc=LabelEncoder()
df.Var_1=var_1_enc.fit_transform(df.Var_1)
var_2_enc=LabelEncoder()
df.Var_2=var_2_enc.fit_transform(df.Var_2)
ID_enc=LabelEncoder()
df.USER_ID=ID_enc.fit_transform(df.USER_ID)
```

```python
X_train, X_test, y_train, y_test=train_test_split(df.drop(["MERCHANT_CATEGORIZED_AS"], axis=1), df.MERCHANT_CATEGORIZED_AS, test_size=.2 ,random_state=25)
```

```python
X_train
```

```python
fit_params={
    "early_stopping_rounds":10, 
    "eval_metric" : 'multi_logloss ', 
    "eval_set" : [(X_test,y_test)],
    'eval_names': ['valid'],
    'verbose': -1,
    'feature_name': 'auto', # that's actually the default
    'categorical_feature': [3,5], # that's actually the default
}
```

```python
clf = LGBMClassifier()
clf.fit(X_train, y_train, **fit_params)
```

```python
predict=clf.predict(X_test)
print(classification_report(y_test, predict))
proba=clf.predict_proba(X_test)
print(f"Log_loss:{log_loss(y_test, proba)}")
```

```python
def light_gbm_model_run(train_x, train_y, validation_x, validation_y, test_x):
    params = {
        "objective" : "multiclass",
        "num_class":11,
        "metric" : "multi_logloss",
        "num_leaves" : 100,
        "learning_rate" : 0.001,
        "bagging_fraction" : 0.6,
        "feature_fraction" : 0.6,
        "bagging_frequency" : 6,
        "bagging_seed" : 42,
        "verbosity" : -1,
        "seed": 42,
        "categorical_feature": [3,5]
    }
    # Given its a regression case, I am using the RMSE as the metric.

    lg_train = lgb.Dataset(train_x, label=train_y)
    lg_validation = lgb.Dataset(validation_x, label=validation_y)
    evals_result_lgbm = {}

    model_light_gbm = lgb.train(params, lg_train, 5000,
                      valid_sets=[lg_train, lg_validation],
                      early_stopping_rounds=100,
                      verbose_eval=150,
                      evals_result=evals_result_lgbm )

    pred_test_light_gbm = model_light_gbm.predict(test_x, num_iteration=model_light_gbm.best_iteration )

    return pred_test_light_gbm, model_light_gbm, evals_result_lgbm
```

```python
# Training and output of LightGBM Model
predictions_test_y_light_gbm, model_lgbm, evals_result = light_gbm_model_run(X_train, y_train, X_test, y_test, X_test)
print('Output of LightGBM Model training..')
```

```python
gain_light_gbm = model_lgbm.feature_importance('gain')
feature_imp_light_gbm = pd.DataFrame({'feature': model_lgbm.feature_name(),
                                      'split': model_lgbm.feature_importance('split'),
                                      'gain': 100 * gain_light_gbm / gain_light_gbm.sum()
}).sort_values('gain', ascending=False)
print(feature_imp_light_gbm)
```
