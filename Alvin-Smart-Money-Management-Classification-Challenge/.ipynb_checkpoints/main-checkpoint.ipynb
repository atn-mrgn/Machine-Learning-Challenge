```python
import warnings
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

warnings.filterwarnings('ignore')
```

```python
extra = pd.read_csv("extra_data.csv")
df = pd.read_csv("Train.csv")
```

```python
plt.figure(figsize=(18,6))
plt.subplot(1,2,1)
sns.barplot((df.isna().sum()*100/df.shape[0]).index, (df.isna().sum()*100/df.shape[0]).values, alpha=0.8)
plt.xticks(rotation=90)
plt.title("Train Data")
plt.subplot(1,2,2)
sns.barplot((extra.isna().sum()*100/extra.shape[0]).index, (extra.isna().sum()*100/extra.shape[0]).values, alpha=0.8)
plt.xticks(rotation=90)
plt.title("Extra Data")

plt.suptitle("Missing Values Percentage", fontsize=18, font="arial")
```
